# Cyber-RAG-Agents : Notebook

This notebook runs the full 3-agent pipeline (Guard → Retriever → Generator)
against the project's test query set and displays every result inline, so
the system's behaviour can be checked and reviewed without re-running the
CLI manually for each question.

**Before running this notebook:**
1. Install dependencies: `pip install -r requirements.txt`
2. Put your source documents in `data/raw_docs/`
3. Build the vector store once: `python -m src.ingestion.build_vectorstore`

Then run all cells below (Kernel → Restart & Run All).

## 1. Setup

```python
import sys
import json
from pathlib import Path

# Make the project root importable (this notebook lives in notebooks/)
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

from src.orchestrator import run_pipeline
from src.config import VECTORSTORE_DIR

print(f"Project root: {PROJECT_ROOT}")
print(f"Vector store path: {VECTORSTORE_DIR}")
```

```python
# Sanity check: make sure the vector store has been built before running queries.
if not VECTORSTORE_DIR.exists() or not any(VECTORSTORE_DIR.iterdir()):
    raise RuntimeError(
        "Vector store not found or empty at "
        f"{VECTORSTORE_DIR}.\n"
        "Run this first from the project root:\n"
        "    python -m src.ingestion.build_vectorstore"
    )
print("Vector store found — ready to run queries.")
```

## 2. Load test queries

```python
queries_path = PROJECT_ROOT / "tests" / "test_queries.json"
with open(queries_path, "r", encoding="utf-8") as f:
    queries = json.load(f)

print(f"Loaded {len(queries)} test queries:")
for i, q in enumerate(queries, start=1):
    print(f"  {i}. {q}")
```

## 3. Run the pipeline on every query

```python
from IPython.display import display, Markdown

results = []

for i, question in enumerate(queries, start=1):
    state = run_pipeline(question)
    results.append({
        "question": question,
        "in_scope": state.get("is_in_scope"),
        "answer": state.get("answer"),
        "sources": state.get("sources", []),
    })

    scope_label = "IN SCOPE" if state.get("is_in_scope") else "OUT OF SCOPE"
    sources_str = ", ".join(state.get("sources", [])) or "—"

    display(Markdown(
        f"### {i}. {question}\n"
        f"**Scope:** {scope_label}  \n"
        f"**Sources:** {sources_str}\n\n"
        f"**Answer:**\n\n{state.get('answer')}\n\n---"
    ))
```

## 4. Summary table

```python
import pandas as pd

df = pd.DataFrame([
    {
        "question": r["question"],
        "in_scope": r["in_scope"],
        "num_sources": len(r["sources"]),
        "sources": ", ".join(r["sources"]),
        "answer_preview": (r["answer"][:120] + "...") if r["answer"] and len(r["answer"]) > 120 else r["answer"],
    }
    for r in results
])

pd.set_option("display.max_colwidth", None)
df
```

## 5. Save results

Writes the same `outputs/test_results.json` file produced by
`tests/run_tests.py`

```python
output_path = PROJECT_ROOT / "outputs" / "test_results.json"
output_path.parent.mkdir(parents=True, exist_ok=True)

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(results, f, indent=2, ensure_ascii=False)

print(f"Saved {len(results)} results to {output_path}")
```

## 6. Inspect retrieved chunks (faithfulness audit)

Check whether a specific claim in an answer is genuinely grounded in the
retrieved context, or whether the model may have added something from its
own pretrained knowledge despite the grounding instruction.

```python
from src.orchestrator import run_pipeline

# Change this to any question or incident description you want to audit
inspect_question = "How does a zero-day vulnerability differ from a known CVE?"

inspect_state = run_pipeline(inspect_question)

print(f"Question: {inspect_question}")
print(f"In scope: {inspect_state.get('is_in_scope')}")
print(f"Chunks retrieved: {len(inspect_state.get('chunks', []))}\n")

for i, chunk in enumerate(inspect_state.get('chunks', []), start=1):
    print(f"--- Chunk {i} — source: {chunk.metadata.get('source', 'unknown')} ---")
    print(chunk.page_content)
    print()

print("=== DRAFT answer (before grounding check) ===")
print(inspect_state.get('draft_answer'))
print()
print("=== FINAL answer (after grounding check) ===")
print(inspect_state.get('answer'))
```
